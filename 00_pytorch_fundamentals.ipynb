TENSOR DATATYPES

NOTE : Tensor datatypes is one of the 3 big errors you will run into with Pytorch and Deep Learning

1. Tensors not right datatype
2. Tensors not right shape
3. Tensors not on the right device


```python
# Default datatype for tensors is float32
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=None, # defaults to None, which is torch.float32 or whatever datatype is passed
                               device=None, # defaults to None, which uses the default tensor type
                               requires_grad=False) # if True, operations performed on the tensor are recorded

float_32_tensor.shape, float_32_tensor.dtype, float_32_tensor.device
```

```python
float_32_tensor.dtype
```

```python
#Converting to float32 to float16
float_16_tensor = float_32_tensor.type(torch.float16)
print(float_16_tensor)
```

```python
float_16_tensor*float_32_tensor
```

```python
int_32_tensor = torch.tensor([3,5,9],dtype=torch.int32)
print(int_32_tensor)
```

```python
int_32_tensor*float_32_tensor
```

MATRIX MULTIPLICATION

Two main ways of performing multiplication in neural networks and deep learning:
1. Element-wise mul
2. Matrix mul (dot product)

The main two rules for matrix multiplication to remember are:

The inner dimensions must match:

(3, 2) @ (3, 2) won't work

(2, 3) @ (3, 2) will work

(3, 2) @ (2, 3) will work

The resulting matrix has the shape of the outer dimensions:

(2, 3) @ (3, 2) -> (2, 2)
(3, 2) @ (2, 3) -> (3, 3)
|
Note: "@" in Python is the symbol for matrix multiplication.

```python
# Element wise multiplication
tensor = torch.tensor([1,2,3])
print(tensor, "*", tensor)
print(f"Equals: {tensor * tensor}")
```

```python
# Matrix multiplication
%%time
torch.matmul(tensor,tensor)
```
